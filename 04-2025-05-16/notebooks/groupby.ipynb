#### `groupby()`-method + aggregation functions

#### Syntax

`df.groupby([Col1,Col2,...])[[Col1,Col2,...]].aggregation_function()`

`df.groupby([group by these columns])[[output these columns]].aggregation_function()`

* `sum()`: Calculates the sum of the group values.  
* `mean()`: Calculates the average of the group values.  
* `median()`: Calculates the median of the group values.  
* `min()`: Determines the minimum of the group values.  
* `max()`: Determines the maximum of the group values.  
* `count()`: Counts the number of non-NA values in each group.  
* `size()`: Returns the number of elements in each group, including NA values.  
* `std()`: Calculates the standard deviation of the group values.  
* `var()`: Calculates the variance of the group values.  
* `agg()`: Applies a custom aggregation function to the groups.
