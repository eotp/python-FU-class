The `Pandas` Series Attributes and Methods:

**Attributes:**

* `name`: The name of the Series.  
* `index`: The index of the Series.  
* `values`: The values of the Series as a NumPy array.  
* `dtype`: The data type of the values in the Series.  
* `shape`: The shape (number of rows and columns) of the Series.  
* `size`: The number of elements in the Series.  
* `empty`: Returns True if the Series is empty.  

**Methods:**

* `head()`: Returns the first n rows of the Series (default: 5).  
* `tail()`: Returns the last n rows of the Series (default: 5).  
* `describe()`: Returns a summary of statistics for the Series.  
* `unique()`: Returns an array of unique values in the Series.  
* `nunique()`: Returns the number of unique values in the Series.  
* `value_counts()`: Returns an array showing the frequency of each value in the Series.  
* `sort_values()`: Sorts the Series by its values.  
* `sort_index()`: Sorts the Series by its index.  
* `nlargest()`: Returns the n largest values in the Series.  
* `nsmallest()`: Returns the n smallest values in the Series.  
* `dropna()`: Removes all NaN values from the Series.  
* `fillna()`: Fills all NaN values in the Series with a specified value.  
* `replace()`: Replaces a specific value in the Series with another value.  
* `apply()`: Applies a function to each value in the Series.  
* `map()`: Replaces each value in the Series using a given mapping.  
* `isna()`: Returns a boolean Series with True for each element that is NaN.  
* `isnull()`: Returns a boolean Series with True for each element that is NaN.

The `Pandas` DataFrame Attributes and Methods:

**Attributes:**

* `index`: The index of the DataFrame.  
* `columns`: The columns of the DataFrame.  
* `values`: The values of the DataFrame as a NumPy array.  
* `shape`: The shape (number of rows and columns) of the DataFrame.  
* `size`: The number of elements in the DataFrame.  
* `dtypes`: The data types of the columns in the DataFrame.  
* `empty`: Returns True if the DataFrame is empty.  

**Methods:**

* `head()`: Returns the first n rows of the DataFrame (default: 5).  
* `tail()`: Returns the last n rows of the DataFrame (default: 5).  
* `describe()`: Returns a summary of statistics for the DataFrame.  
* `info()`: Returns a summary of information about the DataFrame.  
* `transpose()`: Transposes the DataFrame.  
* `sort_values()`: Sorts the DataFrame by the values in one or more columns.  
* `sort_index()`: Sorts the DataFrame by its index.  
* `nlargest()`: Returns the n largest values in the DataFrame.  
* `nsmallest()`: Returns the n smallest values in the DataFrame.  
* `drop()`: Removes one or more columns or rows from the DataFrame.  
* `dropna()`: Removes all rows from the DataFrame that contain NaN values.  
* `fillna()`: Fills all NaN values in the DataFrame with a specified value.  
* `replace()`: Replaces a specific value in the DataFrame with another value.  
* `apply()`: Applies a function to each column or row in the DataFrame.  
* `applymap()`: Applies a function to each individual element in the DataFrame.  
* `groupby()`: Groups the DataFrame by one or more columns.  
* `pivot_table()`: Creates a pivot table from the DataFrame.  
* `join()`: Joins two DataFrames based on a shared index or column.  
* `merge()`: Merges two DataFrames based on shared columns.  
* `query()`: Filters the DataFrame based on a given query.
