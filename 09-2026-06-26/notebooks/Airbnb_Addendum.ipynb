```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
```

```python
airbnb_NY = pd.read_csv("../data/Airbnb_NYC.csv")
```

#### Solution $1$ using `loc` method:

```python
for borough in airbnb_NY["Boroughs"].unique():
    top_10 = airbnb_NY.loc[airbnb_NY["Boroughs"] == borough].sort_values("Reviews30d", ascending=False)[:10]
    print(borough, top_10["Price"].mean())
```

#### Solution $2$ using `groupby` method:

```python
def return_groupby_values(x):
    return x
```

```python
airbnb_NY.groupby(['Boroughs'])[['Boroughs', 'Reviews30d', 'Price']].apply(return_groupby_values)
```

#### `lambda` function

```python
df_ratings = airbnb_NY.groupby(['Boroughs'])[['Boroughs', 'Reviews30d', 'Price']].apply(lambda x: x)

df_ratings
```

```python
df_ratings_boroughs = df_ratings.loc[df_ratings['Boroughs']=='Bronx']

df_ratings_boroughs
```

```python
df_ratings_boroughs_top_ten = df_ratings_boroughs.sort_values("Reviews30d", ascending=False)[:10]

df_ratings_boroughs_top_ten
```
